# Analisis Berdasarkan Perangkat dan Saluran

```python
import matplotlib.pyplot as plt

# Menghitung frekuensi penggunaan setiap perangkat dan saluran
device_counts = df['device'].value_counts()
channel_counts = df['channel'].value_counts()

# Plot perangkat yang sering digunakan
plt.figure(figsize=(10, 5))
device_counts.plot(kind='bar', color='skyblue', title='Frekuensi Penggunaan Perangkat')
plt.xlabel('Perangkat')
plt.ylabel('Jumlah Transaksi')
plt.show()

# Plot saluran transaksi yang sering digunakan
plt.figure(figsize=(10, 5))
channel_counts.plot(kind='bar', color='lightgreen', title='Frekuensi Penggunaan Saluran')
plt.xlabel('Saluran')
plt.ylabel('Jumlah Transaksi')
plt.show()

# Menghitung proporsi penipuan berdasarkan perangkat dan saluran
device_fraud = df.groupby('device')['is_fraud'].mean()
channel_fraud = df.groupby('channel')['is_fraud'].mean()

# Plot proporsi penipuan untuk setiap perangkat
plt.figure(figsize=(10, 5))
device_fraud.plot(kind='bar', color='salmon', title='Proporsi Penipuan Berdasarkan Perangkat')
plt.xlabel('Perangkat')
plt.ylabel('Proporsi Penipuan')
plt.show()

# Plot proporsi penipuan untuk setiap saluran
plt.figure(figsize=(10, 5))
channel_fraud.plot(kind='bar', color='orange', title='Proporsi Penipuan Berdasarkan Saluran')
plt.xlabel('Saluran')
plt.ylabel('Proporsi Penipuan')
plt.show()
```

Berdasarkan hasil grafik, berikut adalah kesimpulan dari analisis perangkat dan saluran transaksi terkait risiko penipuan:

1. **Perangkat Terbanyak**: Perangkat *Edge* memiliki jumlah transaksi tertinggi, namun selisihnya tidak signifikan dibanding perangkat populer lainnya seperti *iOS App*, *Chrome*, *Android App*, *Firefox*, dan *Safari*. Ini menunjukkan penggunaan perangkat yang beragam di antara pengguna.

2. **Saluran Terbanyak dan Berisiko Rendah**: Sistem *web* adalah saluran yang paling sering digunakan, sementara saluran *pos* memiliki frekuensi penggunaan paling rendah namun proporsi penipuan tertinggi. Hal ini menandakan saluran *pos* sebagai titik lemah keamanan, meski jarang digunakan.

3. **Perangkat Berisiko Tinggi**: Metode pembayaran seperti *chip reader*, *magnetic stripe*, dan *NFC payment* menunjukkan proporsi penipuan tertinggi. Hal ini menunjukkan bahwa metode pembayaran fisik ini mungkin lebih rentan terhadap aktivitas penipuan dibanding perangkat berbasis aplikasi atau web.

4. **Rekomendasi Keamanan**: Fokus peningkatan keamanan perlu diarahkan ke transaksi menggunakan saluran *pos* dan perangkat pembayaran fisik seperti *chip reader*, *magnetic stripe*, dan *NFC payment* untuk memitigasi risiko penipuan.


# Analisis Penipuan Berdasarkan Waktu

```python
# Pastikan kolom timestamp dalam format datetime
df['timestamp'] = pd.to_datetime(df['timestamp'], utc=True)
```

```python
# Ekstraksi informasi hari dan jam dari kolom timestamp
df['transaction_hour'] = df['timestamp'].dt.hour
df['transaction_day'] = df['timestamp'].dt.date
```

```python
# Hitung jumlah transaksi per jam
transactions_per_hour = df.groupby('transaction_hour').size()
print(transactions_per_hour)
```

```python
# Hitung jumlah transaksi per hari
transactions_per_day = df.groupby('transaction_day').size()
print(transactions_per_day)
```

```python
# Filter data untuk transaksi yang ditandai sebagai penipuan
fraud_transactions = df[df['is_fraud'] == True]

# Hitung jumlah transaksi penipuan per jam
fraud_per_hour = fraud_transactions.groupby('transaction_hour').size()
df.head()
```

```python
# Visualisasi jumlah transaksi per jam
plt.figure(figsize=(10, 5))
plt.plot(transactions_per_hour, label='Total Transactions')
plt.plot(fraud_per_hour, label='Fraud Transactions', color='red')
plt.xlabel('Hour of Day')
plt.ylabel('Number of Transactions')
plt.title('Transaction and Fraud Patterns by Hour')
plt.legend()
plt.show()
```

Dari grafik tersebut, beberapa kesimpulan awal yang dapat diambil:

1. **Pola Transaksi Per Jam**:
   - Terlihat adanya fluktuasi dalam jumlah total transaksi sepanjang hari, dengan puncak transaksi tertinggi terjadi sekitar pukul 15.00 hingga 18.00. Ini menunjukkan bahwa waktu sore adalah jam sibuk untuk transaksi.
   - Setelah pukul 18.00, jumlah transaksi mulai menurun.

2. **Proporsi Penipuan**:
   - Jumlah transaksi penipuan (garis merah) tetap relatif rendah dibandingkan dengan total transaksi sepanjang hari.
   - Ada sedikit peningkatan transaksi penipuan pada jam-jam awal dan pagi hari, tetapi umumnya tetap stabil dan tidak signifikan dibandingkan dengan total transaksi.

3. **Kaitan Risiko dengan Waktu**:
   - Waktu puncak transaksi tidak selalu memiliki proporsi penipuan yang lebih tinggi, menunjukkan bahwa peningkatan transaksi tidak secara langsung berkorelasi dengan peningkatan risiko penipuan.
   - Waktu dengan sedikit transaksi cenderung menunjukkan tingkat penipuan yang lebih tinggi secara relatif, mungkin karena adanya aktivitas mencurigakan di jam-jam tertentu. 


# Analisis Berdasarkan Kategori Merchant

```python
# Persentase transaksi penipuan
fraud_rate = df['is_fraud'].mean() * 100
print(f"Fraud rate: {fraud_rate:.2f}%")
```

```python
# Persentase transaksi penipuan di merchant berisiko tinggi
high_risk_fraud = df[df['high_risk_merchant'] == True]['is_fraud'].mean() * 100
print(f"Fraud rate for high-risk merchants: {high_risk_fraud:.2f}%")

# Analisis jarak dari rumah untuk transaksi penipuan
fraud_distance = df[df['is_fraud'] == True]['distance_from_home'].value_counts()
non_fraud_distance = df[df['is_fraud'] == False]['distance_from_home'].value_counts()
```

```python
# Distribusi merchant category untuk transaksi fraud
fraud_by_category = df[df['is_fraud'] == True]['merchant_category'].value_counts()
print(fraud_by_category)
```

```python
# Hitung Total Transaksi dan Rata-rata Pengeluaran per Kategori
total_transaksi = df.groupby('merchant_category')['transaction_id'].count()
rata_pengeluaran = df.groupby('merchant_category')['amount'].mean()

# Gabungkan hasil
kategori_stats = pd.DataFrame({
    'Total Transaksi': total_transaksi,
    'Rata-rata Pengeluaran': rata_pengeluaran
})
# Tampilkan 5 baris pertama dari DataFrame
kategori_stats.head()
```

```python
# Hitung Jumlah Penipuan per Kategori
penipuan_per_kategori = df[df['is_fraud'] == 1].groupby('merchant_category')['is_fraud'].count()
# Gabungkan dengan data statistik transaksi
kategori_stats = kategori_stats.join(penipuan_per_kategori.rename('Total Penipuan'), how='left', rsuffix='_penipuan').fillna(0)
```

```python
kategori_stats['Rasio Penipuan (%)'] = (kategori_stats['Total Penipuan'] / kategori_stats['Total Transaksi']) * 100
```

```python
# Visualisasi Data
fig, ax1 = plt.subplots(figsize=(10, 5))

# Total Transaksi dan Rata-rata Pengeluaran
kategori_stats[['Total Transaksi', 'Rata-rata Pengeluaran']].plot(kind='bar', ax=ax1, color=['blue', 'green'])
ax1.set_ylabel("Jumlah Transaksi & Pengeluaran (dalam rata-rata)", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")

# Rasio Penipuan
ax2 = ax1.twinx()
kategori_stats['Rasio Penipuan (%)'].plot(kind='line', ax=ax2, color="red", marker='o', linestyle='--')
ax2.set_ylabel("Rasio Penipuan (%)", color="red")
ax2.tick_params(axis="y", labelcolor="red")

plt.title("Analisis Jumlah Transaksi, Pengeluaran, dan Risiko Penipuan per Kategori Merchant")
plt.show()
```

Berdasarkan grafik analisis jumlah transaksi, pengeluaran, dan risiko penipuan per kategori merchant, berikut adalah beberapa informasi yang dapat disimpulkan:

1. **Jumlah Transaksi**: Hampir semua kategori merchant memiliki jumlah transaksi yang tinggi, sekitar 800.000 transaksi, seperti yang ditunjukkan oleh batang biru di grafik.

2. **Rata-rata Pengeluaran**: Rata-rata pengeluaran (batang hijau) per kategori merchant tampak jauh lebih rendah dibandingkan dengan jumlah transaksi. Ini menunjukkan bahwa meskipun jumlah transaksi tinggi, rata-rata pengeluaran pada masing-masing kategori relatif rendah.

3. **Rasio Penipuan**: Risiko penipuan (garis merah dengan titik) bervariasi di antara kategori merchant, tetapi tetap berada dalam rentang yang relatif sempit (sekitar 19.94% hingga 20.02%). Kategori "Travel" menunjukkan rasio penipuan tertinggi, sedikit di atas 20%, sementara kategori lain cenderung lebih rendah.

4. **Korelasi Potensial**: Tidak ada hubungan yang langsung terlihat antara jumlah transaksi atau rata-rata pengeluaran dengan rasio penipuan. Misalnya, kategori dengan transaksi dan pengeluaran tinggi seperti "Education" dan "Entertainment" memiliki rasio penipuan yang mirip dengan kategori lain, sedangkan "Travel" memiliki rasio yang lebih tinggi meskipun rata-rata pengeluarannya juga relatif rendah.

Secara keseluruhan, grafik ini menunjukkan bahwa meskipun jumlah transaksi dan pengeluaran bervariasi di setiap kategori, risiko penipuan cenderung berada dalam kisaran yang sempit namun lebih tinggi pada kategori tertentu seperti "Travel".
